```python
from sqlalchemy import create_engine
from urllib.parse import quote_plus

# SQL Server connection details
server = r".\SQLEXPRESS"
database = "supermarketAI"
driver = "ODBC Driver 18 for SQL Server"

# Create connection string
connection_string = (
    f"DRIVER={{{driver}}};"
    f"SERVER={server};"
    f"DATABASE={database};"
    "Trusted_Connection=yes;"
    "TrustServerCertificate=yes;"
)

# Encode the connection string
connection_url = (
    "mssql+pyodbc:///?odbc_connect="
    + quote_plus(connection_string)
)

# Create SQLAlchemy engine
engine = create_engine(connection_url)

print("SQLAlchemy engine created successfully!")
```

```python
import pandas as pd

# Load the previously cleaned dataset
df_ai = pd.read_csv("supermarket_ai_data.csv")

# Verify the dataset
print("Dataset loaded successfully!")
print("Rows:", df_ai.shape[0])
print("Columns:", df_ai.shape[1])

df_ai.head()
```

```python
# Import the cleaned dataset into SQL Server

df_ai.to_sql(
    name="SupermarketSales",
    con=engine,
    if_exists="replace",
    index=False
)

print("Dataset imported successfully!")
```

```python
result = pd.read_sql(
    "SELECT COUNT(*) AS TotalRecords FROM SupermarketSales",
    engine
)

print(result)
```
