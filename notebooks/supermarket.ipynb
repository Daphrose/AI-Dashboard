```python
# Calculate revenue using Sales Price and Quantity
df["Calculated Revenue"] = df["Sales Price"] * df["Quantity"]

# Compare calculated revenue with existing revenue
df["Revenue Difference"] = (
    df["Revenue"] - df["Calculated Revenue"]
)

# Display the first 10 rows
df[[
    "Sales Price",
    "Quantity",
    "Discount",
    "Revenue",
    "Calculated Revenue",
    "Revenue Difference"
]].head(10)
```

```python
# Calculate profit using Revenue and COGS
df["Calculated Profit"] = df["Revenue"] - df["COGS"]

# Compare with existing Profit
df["Profit Difference"] = (
    df["Profit"] - df["Calculated Profit"]
)

df[[
    "Revenue",
    "COGS",
    "Profit",
    "Calculated Profit",
    "Profit Difference"
]].head(10)
```

```python
# Count rows where revenue doesn't match Sales Price * Quantity
(df["Revenue Difference"] != 0).sum()
```

```python
# Count matching and mismatching revenue records
df["Revenue Difference"].value_counts()
```

```python
# Display records with revenue mismatches
df[df["Revenue Difference"] != 0][
    [
        "Product",
        "Sales Price",
        "Quantity",
        "Discount",
        "Revenue",
        "Calculated Revenue",
        "Revenue Difference"
    ]
].head(20)
```

```python
# Calculate revenue after subtracting discount
df["Revenue After Discount"] = (
    df["Sales Price"] * df["Quantity"] - df["Discount"]
)

# Compare calculated revenue with actual revenue
df["Discount Difference"] = (
    df["Revenue"] - df["Revenue After Discount"]
)

# Count matching records
(df["Discount Difference"].abs() < 0.01).sum()
```

```python
# Count how many records still don't match
(df["Discount Difference"].abs() >= 0.01).sum()
```

```python
# Calculate profit from Revenue and COGS
df["Calculated Profit"] = df["Revenue"] - df["COGS"]

# Calculate the difference
df["Profit Difference"] = (
    df["Profit"] - df["Calculated Profit"]
)

# Count records that match
(df["Profit Difference"].abs() < 0.01).sum()
```

```python
# Count records that do not match
(df["Profit Difference"].abs() >= 0.01).sum()
```

```python
# Calculate COGS from Unit Cost and Quantity
df["Calculated COGS"] = df["Unit Cost"] * df["Quantity"]

# Compare calculated COGS with recorded COGS
df["COGS Difference"] = (
    df["COGS"] - df["Calculated COGS"]
)

# Count matching records
(df["COGS Difference"].abs() < 0.01).sum()
```

```python
# Count records that don't match
(df["COGS Difference"].abs() >= 0.01).sum()
```

```python
# Get summary statistics for Discount
df["Discount"].describe()
```

```python
# Check the smallest and largest discounts
df["Discount"].min(), df["Discount"].max()
```

```python
# Calculate gross sales before discount
df["Gross Sales"] = df["Sales Price"] * df["Quantity"]

# Calculate discount percentage
df["Discount Percentage"] = (
    df["Discount"] / df["Gross Sales"]
) * 100

# View the first 10 records
df[[
    "Sales Price",
    "Quantity",
    "Gross Sales",
    "Discount",
    "Discount Percentage",
    "Revenue"
]].head(10)
```

```python
(df["Discount"] > df["Gross Sales"]).sum()
```

```python
# Calculate the difference between gross sales and revenue
df["Revenue Reduction"] = df["Gross Sales"] - df["Revenue"]

# Compare the reduction with the recorded discount
df["Reduction Difference"] = (
    df["Revenue Reduction"] - df["Discount"]
)

# Display the first 10 records
df[[
    "Product",
    "Gross Sales",
    "Discount",
    "Revenue",
    "Revenue Reduction",
    "Reduction Difference"
]].head(10)
```

```python
# Count records where revenue reduction equals discount
(
    df["Reduction Difference"].abs() < 0.01
).sum()
```

```python
# Summarize revenue reduction patterns
df["Revenue Reduction"].describe()
```

```python
# Revenue equals Gross Sales
(df["Revenue Reduction"] == 0).sum()
```

```python
# Revenue is lower than Gross Sales
(df["Revenue Reduction"] > 0).sum()
```

```python
# Revenue is higher than Gross Sales
(df["Revenue Reduction"] < 0).sum()
```

```python
# Count all revenue patterns
revenue_equal = (df["Revenue Reduction"] == 0).sum()

revenue_lower = (df["Revenue Reduction"] > 0).sum()

revenue_higher = (df["Revenue Reduction"] < 0).sum()

print("Revenue equals Gross Sales:", revenue_equal)
print("Revenue is lower:", revenue_lower)
print("Revenue is higher:", revenue_higher)

print("Total classified:", revenue_equal + revenue_lower + revenue_higher)

print("Total dataset rows:", len(df))
```

```python
# Summarize revenue reduction by product
product_reduction = (
    df.groupby("Product")["Revenue Reduction"]
    .agg(["count", "mean", "max"])
    .sort_values("max", ascending=False)
)

product_reduction.head(10)
```

```python
# Recalculate Revenue Reduction from the current data
df["Revenue Reduction"] = (
    df["Sales Price"] * df["Quantity"] - df["Revenue"]
)

# Check dataset size
print("Dataset rows:", len(df))

# Count revenue patterns
print("\nRevenue pattern counts:")

print(
    df["Revenue Reduction"]
    .gt(0)
    .sum(),
    "Revenue lower than Gross Sales"
)

print(
    df["Revenue Reduction"]
    .eq(0)
    .sum(),
    "Revenue equal to Gross Sales"
)

print(
    df["Revenue Reduction"]
    .lt(0)
    .sum(),
    "Revenue higher than Gross Sales"
)

# Verify total classified records
print(
    "Total classified:",
    (
        df["Revenue Reduction"].gt(0).sum()
        + df["Revenue Reduction"].eq(0).sum()
        + df["Revenue Reduction"].lt(0).sum()
    )
)

# Recalculate product-level statistics
product_reduction = (
    df.groupby("Product")["Revenue Reduction"]
    .agg(["count", "mean", "max"])
    .sort_values("max", ascending=False)
)

print("\nProduct-level revenue reduction:")
display(product_reduction)
```

```python
# Recalculate Revenue Reduction using the current DataFrame
df["Revenue Reduction"] = (
    df["Sales Price"] * df["Quantity"] - df["Revenue"]
)

# Verify dataset size
print("Dataset shape:", df.shape)

# Recalculate revenue pattern counts
revenue_counts = {
    "Revenue lower": (df["Revenue Reduction"] > 0).sum(),
    "Revenue equal": (df["Revenue Reduction"] == 0).sum(),
    "Revenue higher": (df["Revenue Reduction"] < 0).sum()
}

print("\nRevenue pattern counts:")
print(revenue_counts)

print("\nTotal classified:", sum(revenue_counts.values()))

# Verify product-level statistics
product_check = (
    df.groupby("Product")["Revenue Reduction"]
    .agg(["count", "min", "mean", "max"])
)

display(product_check)

print(
    "\nTotal product records:",
    product_check["count"].sum()
)
```
